```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
products_df = spark.table(
    "ecommerce_lakehouse.silver.products_clean"
)
```

```python
dim_products_df = products_df.select(

    col("product_id"),

    col("product_category_name"),

    col("weight_category"),

    col("product_volume_cm3"),

    col("product_weight_g")

).dropDuplicates()
```

```python
(
    dim_products_df.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(
            "ecommerce_lakehouse.gold.dim_products"
        )
)
```

```python
%sql
SELECT *
FROM ecommerce_lakehouse.gold.dim_products
LIMIT 10;
```
