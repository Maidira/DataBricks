```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
customers_df = spark.table(
    "ecommerce_lakehouse.silver.customers_clean"
)
```

```python
dim_customers_df = customers_df.select(

    col("customer_id"),

    col("customer_unique_id"),

    col("customer_city"),

    col("customer_state")

).dropDuplicates()
```

```python
(
    dim_customers_df.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(
            "ecommerce_lakehouse.gold.dim_customers"
        )
)
```

```python
%sql
SELECT *
FROM ecommerce_lakehouse.gold.dim_customers
LIMIT 10;
```
