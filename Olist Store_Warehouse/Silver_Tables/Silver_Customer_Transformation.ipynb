```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
bronze_customers_df = spark.table(
    "ecommerce_lakehouse.bronze.customers_raw"
)
```

```python
silver_customers_df = bronze_customers_df \
    .dropDuplicates(["customer_id"]) \
    .withColumn(
        "customer_city",
        initcap(col("customer_city"))
    ) \
    .withColumn(
        "customer_state",
        upper(col("customer_state"))
    )
```

```python
(
    silver_customers_df.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(
            "ecommerce_lakehouse.silver.customers_clean"
        )
)
```

```python
%sql
SELECT *
FROM ecommerce_lakehouse.silver.customers_clean
LIMIT 10;
```
